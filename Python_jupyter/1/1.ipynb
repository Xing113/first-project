### 5.ndarray的运算

```python
# 算数的运算
#一维数组
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a ** 2)
```

```python
c = [1,2,3]
d = [4,5,6]
print(a + b)
for i in range(len(c)):
    d[i] = d[i] + c[i]
print(d)
```

```python
# 算数的运算
#二维数组
a = np.array([[1,2,3],[1,2,3]])
b = np.array([[4,5,6],[4,5,6]])
print(a + b)
print(a + 3)
print(a * b)
print(a * 3)
```

```python
#广播机制: 1.获取维度、形状 2.是否可以广播
#(满足其一即可)可以广播：1.两个数组在该维度的大小相同
#        2.其中一方在该维度大小为 1，会被扩展
#        3.缺失的维度视为 1
a = np.array([1,2,3])  #1*3
print(a.shape)
b = np.array([[4],[5],[6]])  #3*1
print(b.shape)
print(a+b)
```

```python
#分位数
arr = np.random.randint(0,100,5)
print(arr)
```

|------------------|

 37  50  69  71  77

 中位数：69

 分位数：看这组数中所在位置（0%-100%）的对应数值。

5个数分了4段，找到该百分比所对应的段位，用这一段的数值范围×这个数在该段的百分比+该段起始值


```python
print(np.median(arr))
print(np.percentile(arr,50)) #50%——>中位数
print(np.percentile(arr,20))
#4*0.2=0.8 -->第一段  ；  （50-37）*0.8=10.4   ； 10.4+37=47.4
print(np.percentile(arr,80))
#4*0.8=3.2 -->第4段  ；  （77-71）*0.2=1.2   ； 1.2+71=72.2
print(np.percentile(arr,60))
#4*0.6=2.4 -->第3段  ；  （71-69）*0.4=0.8   ； 0.8+69=69.8
```

## numpy练习题

1---温度数据分析

[28,30,29,31,32,30,29]

```python
#平均气温
import numpy as np
arr = np.array([28,30,29,31,32,30,29])
print('平均气温：','%.3f'%np.mean(arr))
print('最高气温：',np.max(arr))
print('最低气温：',np.min(arr))
print('气温超过30的天数：',len(arr[arr>30]))

a = np.where(arr>30)[0]  # 返回一维数组 [4, 5]
print(f'第{a+1}天超过30')

a = np.argwhere(arr>30)
print(f'第{a+1}天超过30')

a = np.where(arr>30)[0] + 1  # [, 5]
days_str = '、'.join(map(str, a))  # 变成 "4、5"  #map(str, a)：把数组的每个元素转成字符串  #'、'.join(...)：用顿号连接所有字符串
print(f'第{days_str}天超过30')
# 输出：第4、5天超过30

b = np.sum(np.where(arr>30,1,0))
print('气温超过30的天数：',b)
```
